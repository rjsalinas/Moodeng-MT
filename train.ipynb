```python
import pandas as pd

df = pd.read_csv('filipino_english_parallel_corpus.csv')

print(df.head(3))
```

```python
from sklearn.model_selection import train_test_split
```

```python

train, val = train_test_split(df, test_size=0.1, random_state=42)

train["preprocessed_text"].to_csv("corpus-parallel-txt/train.tl", index=False, header=False)
train["english_translation"].to_csv("corpus-parallel-txt/train.en", index=False, header=False)

val["preprocessed_text"].to_csv("corpus-parallel-txt/val.tl", index=False, header=False)
val["english_translation"].to_csv("corpus-parallel-txt/val.en", index=False, header=False)

# df["preprocessed_text"].to_csv("corpus-parallel-txt/corpus.tl", index=False, header=False)
# df["english_translation"].to_csv("corpus-parallel-txt/corpus.en", index=False, header=False)
```
