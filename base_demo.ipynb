```python
from transformers import MBart50TokenizerFast, MBartForConditionalGeneration

# Load model and tokenizer
model_name = "facebook/mbart-large-50-many-to-many-mmt"
tokenizer = MBart50TokenizerFast.from_pretrained(model_name)
model = MBartForConditionalGeneration.from_pretrained(model_name)
```

```python
# Set source language (input)
tokenizer.src_lang = "en_XX"

# Your input text
text = "How are you?"

# Tokenize
encoded = tokenizer(text, return_tensors="pt")

# Generate translation
generated_tokens = model.generate(
    **encoded,
    forced_bos_token_id=tokenizer.lang_code_to_id["es_XX"]  # Target language
)

# Decode
translated_text = tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)[0]

print("Input (English):", text)
print("Output (Spanish):", translated_text)

# Generate Tagalog translation
generated_tokens_tl = model.generate(
    **encoded,
    forced_bos_token_id=tokenizer.lang_code_to_id["tl_XX"]
)

translated_text_tl = tokenizer.batch_decode(generated_tokens_tl, skip_special_tokens=True)[0]

print("Output (Tagalog):", translated_text_tl)
```
